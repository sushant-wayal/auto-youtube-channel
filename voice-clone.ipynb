```python
# Cell 3 — Zero-Shot Voice Cloning Inference

import os
import sys
import torch
import soundfile as sf

# Add cloned repo to python path

sys.path.append("/content/F5-TTS/src")

from f5_tts.api import F5TTS

print("✅ Imports successful")

# -----------------------------

# CONFIG

# -----------------------------

REFERENCE_AUDIO = list(uploaded.keys())[0]

# Exact transcript of uploaded audio

REFERENCE_TEXT = """
Most distributed systems fail silently before they fail visibly.
A retry loop without backoff can destroy a production service.
"""

# New text to generate in YOUR voice

GENERATE_TEXT = """
In the last decade, the narrative around computing has shifted. We've seen Apple Silicon and AWS Graviton dominate headlines, leading many to believe that ARM is inherently 'better' or more efficient than x86. But if you talk to a CPU architect, they'll tell you something different. The Instruction Set Architecture—or ISA—is just the language a chip speaks. It isn't the engine itself. Today, we're deconstructing the myth that ARM’s success is purely due to its Reduced Instruction Set. We’ll look at how modern x86 chips have evolved to act like ARM internally, and why the real performance gains come from micro-architecture and manufacturing nodes, not just the dictionary of commands the processor understands. By the end of this, you'll see that the gap between these architectures is narrower than marketing suggests.
"""

OUTPUT_FILE = "generated_voice.wav"

# -----------------------------

# LOAD MODEL

# -----------------------------

tts = F5TTS()

print("✅ Model Loaded")

# -----------------------------

# GENERATE AUDIO

# -----------------------------

wav, sr, _ = tts.infer(
ref_file=REFERENCE_AUDIO,
ref_text=REFERENCE_TEXT,
gen_text=GENERATE_TEXT,
)

# Save output

sf.write(OUTPUT_FILE, wav, sr)

print(f"✅ Voice generated: {OUTPUT_FILE}")

# -----------------------------

# PLAY AUDIO

# -----------------------------

from IPython.display import Audio

Audio(OUTPUT_FILE)
```
